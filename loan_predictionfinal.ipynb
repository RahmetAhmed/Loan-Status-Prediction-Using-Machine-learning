```python

data['LoanStatus'] = data['LoanStatus'].map({'Accepted':1,'Rejected':0}).astype('int')
```

```python
data.head()
```

```python
#8. Store feature matrix and in X and and response(Target) in vector y
#data.drop('LoanStatus', axis=1)
X = data.drop('LoanStatus', axis=1)
X
y = data['LoanStatus']
y


```

```python
# 9 Feature scaling
cols = ['Customer_No', 'ApplicantIncome', 'LoanAmount', 'LoanTerm']

from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X[cols] = sc.fit_transform(X[cols])
X
```

```python
# 10 Spliting the dataset into training set aand test set and than apply K-fold cross validation
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
import numpy as np
import joblib

model_df = {}  # Initialize the model_df dictionary

def model_val(model, X, y):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"{model} accuracy is {accuracy:.2f}")
    
    scores = cross_val_score(model, X, y, cv=5)
    avg_score = np.mean(scores)
    print(f"{model} Avg cross val score is {avg_score:.2f}")
    
    model_df[str(model)] = round(avg_score * 100, 2)
	
```

```python
# 11 Logistic regression
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model_val(model,X,y)
```

```python

# 12 Hyperparameter tuning for Logistic Regression
from sklearn.model_selection import RandomizedSearchCV
log_rig_grid={ "C":np.logspace(4,-4,20), "solver":['liblinear']}
rs_log_reg=RandomizedSearchCV(LogisticRegression(),param_distributions=log_rig_grid,n_iter=20,cv=5,verbose=True)
rs_log_reg.fit(X,y)
# Run the below code one by one to get to get the output
rs_log_reg.best_score_
```
